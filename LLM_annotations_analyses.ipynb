#### RQ 1_2: Did participants search for data or papers?

```python
# data and paper search

data_search_t1 = {}
data_search_t2 = {}
papers_search_t1 = {}
papers_search_t2 = {}

for i, r in data.iterrows():
    if r['R1_2_data'] == 'x':
        # unguided task
        if r['# of task'] == 1:
            if r['id'] in data_search_t1:
                data_search_t1[r['id']] = data_search_t1[r['id']] + 1
            else:
                data_search_t1[r['id']] = 1
        # guided task
        else:
            if r['id'] in data_search_t2:
                data_search_t2[r['id']] = data_search_t2[r['id']] + 1
            else:
                data_search_t2[r['id']] = 1
    if r['R1_2_papers'] == 'x':
        # unguided task
        if r['# of task'] == 1:
            if r['id'] in papers_search_t1:
                papers_search_t1[r['id']] = papers_search_t1[r['id']] + 1
            else:
                papers_search_t1[r['id']] = 1
        # guided task
        else:
            if r['id'] in papers_search_t2:
                papers_search_t2[r['id']] = papers_search_t2[r['id']] + 1
            else:
                papers_search_t2[r['id']] = 1
```

```python
# persons searching for data in unguided task
print(len(data_search_t1))
# persons searching for data in guided task
print(len(data_search_t2))
# persons searching for papers in unguided task
print(len(papers_search_t1))
# persons searching for papers in guided task
print(len(papers_search_t2))

print('___')
# persons searching for data in both task
print(len(data_search_t1.keys() & data_search_t2.keys()))
# persons searching for papers in both task
print(len(papers_search_t1.keys() & papers_search_t2.keys()))

print('___')
# persons searching for data in and papers in unguided task
print(len(data_search_t1.keys() & papers_search_t1.keys()))
# persons searching for data in and papers in unguided task
print(len(data_search_t2.keys() & papers_search_t2.keys()))
```

### RQ 2: Do the interactions between participants and LLMs show characteristics associated with humanisation?

#### RQ 2_1: Do participants' and LLMs' utilised dialogue acts differ in their interactions?

```python
len_chatgpt_unguided = len(data.loc[(data['# of task'] == 1) & (data['LLM'] == 1)])
len_chatgpt_guided = len(data.loc[(data['# of task'] == 2) & (data['LLM'] == 1)]) - 16
len_perplexity_unguided = len(data.loc[(data['# of task'] == 1) & (data['LLM'] == 2)])
len_perplexity_guided = len(data.loc[(data['# of task'] == 2) & (data['LLM'] == 2)]) - 16
```

```python
fields = ['P_statement', 'LLM_statement', 'P_opinion', 'LLM_opinion', 'P_question', 'LLM_question',  'P_answers', 'LLM_answers', 'P_appreciation', 'LLM_appreciation', 'P_apology', 'LLM_apology', 'P_thanking', 'LLM_thanking',   'P_action', 'LLM_action', 'P_offer', 'LLM_offer', 'P_opening', 'LLM_opening', 'P_closing', 'LLM_closing', 'P_hold', 'LLM_hold',  'P_copypaste_phrasing', 'P_abandon']

task_1 = data.loc[data['# of task'] == 1]
task_2 = data.loc[data['# of task'] == 2]

for f in fields:
    # ChatGPT
    ds_t1_c = {}
    ds_t2_c = {}
    
    for i, r in task_1.iterrows():
        if r['LLM'] == 1 and (r[f] == 'x' or r[f] == 'y' or r[f] == 'n' or r[f] == 'my' or r[f] == 'w'):
            if r['id'] in ds_t1_c:
                ds_t1_c[r['id']] = ds_t1_c[r['id']] + 1
            else:
                ds_t1_c[r['id']] = 1
    for i, r in task_2.iterrows():
        if r['LLM'] == 1 and (r[f] == 'x' or r[f] == 'y' or r[f] == 'n' or r[f] == 'my' or r[f] == 'w'):
            if r['id'] in ds_t2_c:
                ds_t2_c[r['id']] = ds_t2_c[r['id']] + 1
            else:
                ds_t2_c[r['id']] = 1
    
    if f[0] == 'P':
        print(f[2].upper() + f[3:] + ' & ')

    # Perplexity
    ds_t1_p = {}
    ds_t2_p = {}
    
    for i, r in task_1.iterrows():
        if r['LLM'] == 2 and (r[f] == 'x' or r[f] == 'y' or r[f] == 'n' or r[f] == 'my' or r[f] == 'w'):
            if r['id'] in ds_t1_p:
                ds_t1_p[r['id']] = ds_t1_p[r['id']] + 1
            else:
                ds_t1_p[r['id']] = 1
    for i, r in task_2.iterrows():
        if r['LLM'] == 2 and (r[f] == 'x' or r[f] == 'y' or r[f] == 'n' or r[f] == 'my' or r[f] == 'w'):
            if r['id'] in ds_t2_p:
                ds_t2_p[r['id']] = ds_t2_p[r['id']] + 1
            else:
                ds_t2_p[r['id']] = 1

    # combination of LLMs
    print(  str(len(ds_t1_c) + len(ds_t1_p))                 + ' (' + str(round(data.loc[(data['# of task'] == 1)][f].count()/(len_chatgpt_unguided + len_perplexity_unguided) * 100, 2)) + '\%) & ' +\
            str(len(ds_t2_c) + len(ds_t2_p))                 + ' (' + str(round(data.loc[(data['# of task'] == 2)][f].count()/(len_chatgpt_guided + len_perplexity_guided) * 100, 2)) + '\%)')

    if f[0] == 'P':
        print(' & ')
    else:
        print('\\\\')
```

```python
# How many yes/no and wh-questions were posed by participants and LLMs, how many yes/no questions were answered
LLM_query_yn = 0
LLM_query_w = 0

part_query_yn = 0
part_query_w = 0

LLM_answer_yn = 0
part_answer_yn = 0

for i, r in data.iterrows():
    if r['P_question'] == 'y' or r['P_question'] == 'my':
        part_query_yn += 1
        
        if r['LLM_answers'] == 'y' or r['LLM_answers'] == 'n':
            LLM_answer_yn += 1

    if r['P_question'] == 'w':
        part_query_w += 1

    if r['LLM_question'] == 'y' or r['LLM_question'] == 'my':
        LLM_query_yn += 1
        
    if r['P_answers'] == 'y' or r['P_answers'] == 'n':
        part_answer_yn += 1

    if r['LLM_question'] == 'w':
        LLM_query_w += 1
        
print('participants, yes/no questions: ' + str(part_query_yn))
print('participants, wh questions: ' + str(part_query_w))

print('participants, yes/no answers: ' + str(part_answer_yn))

print('___')
print('LLMs, yes/no questions: ' + str(LLM_query_yn))
print('LLMs, wh questions: ' + str(LLM_query_w))

print('LLMs, yes/no questions: ' + str(LLM_answer_yn))
```
